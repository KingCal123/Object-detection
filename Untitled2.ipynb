# Introduction to Python and Variables

Python is a versatile, high-level programming language known for its readability and simplicity. When writing programs, we often need to store and manipulate data. This is where **variables** come in.

### What is a Variable?
A variable acts as a container or a label for storing data values. In Python, you don't need to declare a variable's type before using it; Python dynamically infers the type based on the value assigned to it.

### Fundamental Data Types in Python
Python has several built-in fundamental data types:
1. **Integer (`int`)**: Whole numbers, positive or negative, without decimals (e.g., `42`, `-5`).
2. **Floating-point (`float`)**: Numbers that contain one or more decimals (e.g., `19.99`, `-0.5`).
3. **String (`str`)**: Textual data enclosed in single or double quotes (e.g., `"Alice"`, `'Python'`).
4. **Boolean (`bool`)**: Represents one of two values: `True` or `False`.

# Basic Operators and Expressions

Operators are special symbols in Python that carry out arithmetic or logical computation. The value that the operator operates on is called the operand.

### 1. Arithmetic Operators
Arithmetic operators are used to perform common mathematical operations:
*   **Addition (`+`)**: Adds two values (e.g., `5 + 3 = 8`).
*   **Subtraction (`-`)**: Subtracts the right operand from the left operand (e.g., `5 - 3 = 2`).
*   **Multiplication (`*`)**: Multiplies two values (e.g., `5 * 3 = 15`).
*   **Division (`/`)**: Divides the left operand by the right operand, always returning a float (e.g., `5 / 2 = 2.5`).
*   **Floor Division (`//`)**: Divides and rounds down to the nearest whole number (e.g., `5 // 2 = 2`).
*   **Modulo (`%`)**: Returns the remainder of the division (e.g., `5 % 2 = 1`).
*   **Exponentiation (`**`)**: Performs power calculation (e.g., `5 ** 2 = 25`).

### 2. Comparison Operators
Comparison operators compare two values and return a Boolean value (`True` or `False`):
*   **Equal to (`==`)**: Checks if two values are equal.
*   **Not equal to (`!=`)**: Checks if two values are not equal.
*   **Greater than (`>`)** and **Less than (`<`)**
*   **Greater than or equal to (`>=`)** and **Less than or equal to (`<=`)**

### 3. Logical Operators
Logical operators are used to combine conditional statements:
*   **`and`**: Returns `True` if both statements are true.
*   **`or`**: Returns `True` if at least one of the statements is true.
*   **`not`**: Reverses the logical state (turns `True` to `False` and vice versa).

**Reasoning**:
Create a python code cell to demonstrate arithmetic, comparison, and logical operators with clear print statements and explanations.



```python
# --- 1. Arithmetic Operators Demonstration ---
a = 15
b = 4
print("=== Arithmetic Operators ===")
print(f"a = {a}, b = {b}\n")
print(f"Addition (a + b): {a} + {b} = {a + b}")
print(f"Subtraction (a - b): {a} - {b} = {a - b}")
print(f"Multiplication (a * b): {a} * {b} = {a * b}")
print(f"Division (a / b): {a} / {b} = {a / b}  (Notice this always returns a float!)")
print(f"Floor Division (a // b): {a} // {b} = {a // b}  (Rounds down to the nearest integer)")
print(f"Modulo (a % b): {a} % {b} = {a % b}  (Returns the remainder of division)")
print(f"Exponentiation (a ** b): {a} ** {b} = {a ** b}  (15 raised to the power of 4)")
print()

# --- 2. Comparison Operators Demonstration ---
print("=== Comparison Operators ===")
print(f"Is a equal to b? (a == b): {a == b}")
print(f"Is a not equal to b? (a != b): {a != b}")
print(f"Is a greater than b? (a > b): {a > b}")
print(f"Is a less than b? (a < b): {a < b}")
print(f"Is a greater than or equal to b? (a >= b): {a >= b}")
print(f"Is a less than or equal to b? (a <= b): {a <= b}")
print()

# --- 3. Logical Operators Demonstration ---
print("=== Logical Operators ===")
x = True
y = False
print(f"x = {x}, y = {y}\n")
print(f"Logical AND (x and y): {x and y}  (True only if both are True)")
print(f"Logical OR (x or y): {x or y}  (True if at least one is True)")
print(f"Logical NOT (not x): {not x}  (Reverses the value of x)")
print()

# Combining comparisons with logical operators
expr1 = (a > 10) and (b < 5)  # True and True -> True
expr2 = (a < 10) or (b < 5)   # False or True -> True
print("=== Combined Logical Expressions ===")
print(f"Is 'a > 10 and b < 5' true? {expr1}")
print(f"Is 'a < 10 or b < 5' true? {expr2}")
```

# Control Flow: Conditionals and Loops

Control flow is the order in which individual statements, instructions, or function calls of an imperative program are executed or evaluated. In Python, control flow is managed using conditional statements and loops.

### 1. Conditional Statements (`if`, `elif`, `else`)
Conditional statements allow your code to make decisions based on conditions. Unlike many other programming languages, Python uses **indentation** (typically 4 spaces) to define code blocks instead of curly braces `{}` or keywords like `end`.

*   **`if`**: Executes a block of code if its condition evaluates to `True`.
*   **`elif`** (short for else if): Checks another condition if the preceding conditions were `False`.
*   **`else`**: Executes a block of code if all preceding conditions were `False`.

### 2. Loops
Loops are used to repeat a block of code multiple times.

*   **`for` Loop**: Used for iterating over a sequence (such as a list, a tuple, a dictionary, a set, or a string) or a range of numbers. It is ideal when you know beforehand how many times you need to execute the loop.
*   **`while` Loop**: Repeats a block of code as long as a specified condition remains `True`. It is ideal when the number of iterations is not known beforehand.

**Reasoning**:
Create a Python code cell demonstrating conditionals (if-elif-else) for grading, a basic for loop with range, and a simple while loop countdown.



```python
# --- 1. Conditional Statements (if-elif-else) ---
# Let's assign a grade based on a score
score = 85

print("=== Conditional Statements ===")
print(f"Score to evaluate: {score}")

if score >= 90:
    grade = "A"
elif score >= 80:
    grade = "B"
elif score >= 70:
    grade = "C"
elif score >= 60:
    grade = "D"
else:
    grade = "F"

print(f"Resulting Grade: {grade}\n")

# --- 2. For Loop (Iterating over a range) ---
print("=== For Loop (Iterating over a range) ===")
# printing numbers from 1 to 5
for i in range(1, 6):
    print(f"Iteration {i}")
print()

# --- 3. While Loop (Countdown) ---
print("=== While Loop (Countdown) ===")
countdown = 3
while countdown > 0:
    print(f"T-minus {countdown}...")
    countdown -= 1  # Equivalent to countdown = countdown - 1
print("Liftoff!\n")
```

# Data Structures: Lists, Tuples, and Dictionaries

In Python, built-in collection data structures allow us to store groups of data together under a single variable name. The most commonly used structures are lists, tuples, and dictionaries.

### 1. Lists (`list`)
*   **Definition**: A mutable, ordered sequence of elements.
*   **Mutability**: **Mutable**, meaning you can change, add, or remove items after the list is created.
*   **Syntax**: Defined using square brackets `[]` with elements separated by commas.

### 2. Tuples (`tuple`)
*   **Definition**: An immutable, ordered sequence of elements.
*   **Mutability**: **Immutable**, meaning once created, its elements cannot be changed, added, or removed.
*   **Syntax**: Defined using parentheses `()` with elements separated by commas.

### 3. Dictionaries (`dict`)
*   **Definition**: A mutable, unordered collection of key-value pairs.
*   **Mutability**: **Mutable** (keys must be immutable, but values can be changed or added).
*   **Syntax**: Defined using curly braces `{}` containing `key: value` pairs separated by commas.

**Reasoning**:
Create a code cell to demonstrate creation, indexing, slicing, and manipulation for lists, tuples, and dictionaries as specified in the subtask instructions.



```python
# === 1. Lists Demonstration (Mutable & Ordered) ===
print("=== 1. Lists ===")
fruits = ["apple", "banana", "cherry", "date"]
print("Original list:", fruits)

# Accessing elements using indexing (0-based)
print("First element (fruits[0]):", fruits[0])
print("Last element (fruits[-1]):", fruits[-1])

# Accessing elements using slicing
print("Sliced list (fruits[1:3]):", fruits[1:3])

# Appending a new element
fruits.append("elderberry")
print("After append:", fruits)

# Modifying an element
fruits[1] = "blueberry"
print("After modifying index 1:", fruits)
print()

# === 2. Tuples Demonstration (Immutable & Ordered) ===
print("=== 2. Tuples ===")
dimensions = (1920, 1080)
print("Original tuple:", dimensions)

# Accessing elements
print("Width (dimensions[0]):", dimensions[0])
print("Height (dimensions[1]):", dimensions[1])

# Attempting to modify a tuple raises an error
# dimensions[0] = 1280  # This would raise TypeError: 'tuple' object does not support item assignment
print("Note: Attempting to modify dimensions[0] directly results in a TypeError because tuples are immutable.")
print()

# === 3. Dictionaries Demonstration (Mutable Key-Value Pairs) ===
print("=== 3. Dictionaries ===")
user_profile = {
    "username": "johndoe",
    "email": "john@example.com",
    "is_admin": False
}
print("Original dictionary:", user_profile)

# Accessing values using keys
print("Email address:", user_profile["email"])

# Adding a new key-value pair
user_profile["country"] = "Canada"

# Updating an existing key-value pair
user_profile["is_admin"] = True

print("Updated dictionary:", user_profile)
```

# Functions and Variable Scope

Functions are reusable blocks of code designed to perform a specific task. They help organize code, reduce repetition, and make programs easier to maintain.

### 1. Defining and Calling Functions
In Python, you define a function using the `def` keyword, followed by the function name, parentheses `()`, and a colon `:`. The code block inside the function must be indented.

*   **Parameters/Arguments**: Variables listed inside the parentheses in the function definition. They allow you to pass data into the function.
*   **Return Statement**: Used to send a result back to the caller using the `return` keyword. If no return statement is specified, the function returns `None` by default.

### 2. Variable Scope
Scope refers to the region of a program where a variable is accessible.

*   **Local Scope**: Variables created inside a function belong to that function's local scope. They can only be accessed from within that function.
*   **Global Scope**: Variables created outside of any function belong to the global scope. They can be read from anywhere within the program, including inside functions.

**Reasoning**:
Create a code cell illustrating a function definition with parameters and a return value, and demonstrate the practical difference between local and global scope.



```python
# --- 1. Defining and Calling a Function ---
def calculate_rectangle_area(width, height):
    """Calculates and returns the area of a rectangle."""
    area = width * height
    return area

# Calling the function and storing the result
rect_width = 5.0
rect_height = 8.0
calculated_area = calculate_rectangle_area(rect_width, rect_height)

print("=== 1. Function Definition & Return Value ===")
print(f"Width: {rect_width}, Height: {rect_height}")
print(f"Calculated Area: {calculated_area}\n")

# --- 2. Demonstrating Variable Scope ---
global_message = "I am a global variable!"

def scope_demo_function():
    # This variable is local to scope_demo_function
    local_message = "I am a local variable!"
    print("=== Inside the Function ===")
    print("Can access global variable:", global_message)
    print("Can access local variable:", local_message)

# Execute the function to see the print statement outputs
scope_demo_function()
print()

print("=== Outside the Function ===")
print("Can access global variable:", global_message)

try:
    # Attempting to access the local variable outside its function
    print("Can access local variable?:", local_message)
except NameError as e:
    print("Error attempting to access local_message:", e)
    print("Note: local_message is only accessible inside 'scope_demo_function'.")
```

# Basic File I/O and Error Handling

### 1. Working with Files (`with open()`)
In Python, the preferred way to open and interact with files is using the `with` statement, also known as a **context manager**.

*   **Automatic Closure**: The `with open()` pattern ensures that the file is automatically and properly closed after the block of code finishes executing, even if an error occurs inside the block. This prevents resource leaks.
*   **Modes**: Common modes include `'w'` for writing (which overwrites existing content), `'r'` for reading, and `'a'` for appending.

### 2. Error Handling (`try-except`)
Errors detected during execution are called **exceptions**. If they are not handled, the program crashes. Python uses `try` and `except` blocks to handle these exceptions gracefully.

*   **`try` block**: Contains the code that might raise an exception.
*   **`except` block**: Contains the code that executes if an exception is raised in the `try` block. You can target specific errors (such as `FileNotFoundError` or `ValueError`).
*   **`finally` block**: An optional block that runs no matter what, whether an exception was raised or not, often used for clean-up operations.

**Reasoning**:
Create a code cell demonstrating writing text to a file using 'with open', attempting to read a non-existent file with a try-except block to handle FileNotFoundError, and successfully reading 'sample.txt' within a try-except block.



```python
import os

# --- 1. Writing to a File using a Context Manager ---
filename = 'sample.txt'
print("=== 1. Writing to File ===")
with open(filename, 'w') as file:
    file.write("Hello, Python world!\n")
    file.write("This is a demonstration of file I/O with error handling.\n")
    file.write("The 'with' statement handles file closure automatically.\n")
print(f"Successfully wrote content to '{filename}'.\n")

# --- 2. Handling FileNotFoundError ---
non_existent_file = 'ghost_file.txt'
print("=== 2. Attempting to Read a Non-existent File ===")
try:
    with open(non_existent_file, 'r') as file:
        content = file.read()
except FileNotFoundError as e:
    print(f"Caught expected exception: {e}")
    print(f"The file '{non_existent_file}' does not exist, but our program did not crash!\n")

# --- 3. Successfully Reading from the Created File ---
print("=== 3. Successfully Reading standard File ===")
try:
    with open(filename, 'r') as file:
        content = file.read()
        print("File Contents:")
        print(content.strip())
except FileNotFoundError:
    print(f"Error: The file '{filename}' was not found.")
finally:
    # Cleaning up the created sample.txt file after presentation
    if os.path.exists(filename):
        os.remove(filename)
        print(f"\nCleaned up: '{filename}' has been deleted.")
```
