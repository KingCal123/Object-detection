# Python Functions: A Comprehensive Guide

According to the official Python documentation, a **function** is a reusable block of code that perform a specific task. Functions help break our program into smaller and modular chunks. As our program grows larger, functions make it more organized and manageable.

### Key Concepts Covered:
1. **Defining a Function** (`def` keyword)
2. **Docstrings** (Documentation strings)
3. **Arguments and Parameters** (Default, Positional, Keyword, `*args`, and `**kwargs`)
4. **Return Values**
5. **Scope and Lifetime of Variables**
6. **Lambda (Anonymous) Functions**

## 1. Defining and Calling a Function

To define a function in Python, we use the `def` keyword, followed by the function name, parentheses `()`, and a colon `:`. The body of the function is indented.

```python
def greet():
    """This is a simple docstring explaining that this function greets the user."""
    print("Hello! Welcome to Python functions.")

# Calling the function
greet()
```

## 2. Arguments: Positional and Keyword

* **Positional arguments**: Passed to a function in order of their definition.
* **Keyword arguments**: Passed using the format `parameter_name = value`. This allows passing them out of order.

```python
def describe_pet(animal_type, pet_name):
    print(f"I have a {animal_type} named {pet_name}.")

# Positional arguments (order matters)
describe_pet("dog", "Rex")

# Keyword arguments (order does not matter)
describe_pet(pet_name="Whiskers", animal_type="cat")
```

## 3. Default Parameter Values

You can provide default values for parameters. If the caller does not supply an argument for that parameter, the default value is used.

```python
def make_coffee(size, roast="Medium"):
    print(f"Making a {size} coffee with a {roast} roast.")

# Using the default value
make_coffee("Large")

# Overriding the default value
make_coffee("Small", roast="Dark")
```

## 4. Arbitrary Arguments: `*args` and `**kwargs`

* `*args`: Allows a function to accept any number of positional arguments (received as a tuple).
* `**kwargs`: Allows a function to accept any number of keyword arguments (received as a dictionary).

```python
def sum_all(*numbers):
    # numbers is a tuple
    return sum(numbers)

print("Sum of 1, 2, 3:", sum_all(1, 2, 3))
print("Sum of 5, 10, 15, 20:", sum_all(5, 10, 15, 20))

def build_profile(first, last, **user_info):
    # user_info is a dictionary
    user_info['first_name'] = first
    user_info['last_name'] = last
    return user_info

profile = build_profile('Guido', 'van Rossum', location='Netherlands', field='Software')
display(profile)
```

## 5. Lambda Functions (Anonymous Functions)

Lambda functions are small, anonymous functions defined using the `lambda` keyword. They can have any number of arguments but only one expression.

```python
# Standard function
def square(x):
    return x ** 2

# Equivalent Lambda function
square_lambda = lambda x: x ** 2

print("Square using standard function:", square(5))
print("Square using lambda function:", square_lambda(5))

# Practical usage: Sorting lists of tuples with a lambda key
pairs = [(1, 'one'), (2, 'two'), (3, 'three'), (4, 'four')]
pairs.sort(key=lambda pair: pair[1])
print("Sorted by string name:", pairs)
```
